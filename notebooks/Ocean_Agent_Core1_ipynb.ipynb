```python
import numpy as np

# Make a copy so original dataset remains safe
ocean_df = df.copy()

# ---------------------------------------------------
# 1. Current direction from U and V components
# ---------------------------------------------------
ocean_df["current_direction_deg"] = (
    np.degrees(
        np.arctan2(
            ocean_df["current_v_ms"],
            ocean_df["current_u_ms"]
        )
    ) + 360
) % 360


# ---------------------------------------------------
# 2. Chlorophyll transformation
#    Helps reduce effect of very high outliers
# ---------------------------------------------------
ocean_df["chlorophyll_log"] = np.log1p(
    ocean_df["chlorophyll_mean"]
)


# ---------------------------------------------------
# 3. SST deviation from dataset median
#    Useful for suitability-based scoring later
# ---------------------------------------------------
sst_median = ocean_df["sst_c"].median()

ocean_df["sst_deviation"] = (
    ocean_df["sst_c"] - sst_median
).abs()


# ---------------------------------------------------
# 4. Current deviation from dataset median
#    Useful for suitability modelling later
# ---------------------------------------------------
current_median = ocean_df["current_speed_ms"].median()

ocean_df["current_deviation"] = (
    ocean_df["current_speed_ms"] - current_median
).abs()


# ---------------------------------------------------
# 5. Ocean data completeness
# ---------------------------------------------------
core_ocean_features = [
    "sst_c",
    "chlorophyll_mean",
    "current_u_ms",
    "current_v_ms",
    "current_speed_ms"
]

ocean_df["ocean_features_available"] = (
    ocean_df[core_ocean_features]
    .notna()
    .sum(axis=1)
)

ocean_df["ocean_data_completeness"] = (
    ocean_df["ocean_features_available"]
    / len(core_ocean_features)
)


# ---------------------------------------------------
# 6. Check engineered features
# ---------------------------------------------------
engineered_features = [
    "zone_id",
    "region",
    "latitude",
    "longitude",
    "sst_c",
    "chlorophyll_mean",
    "chlorophyll_log",
    "current_u_ms",
    "current_v_ms",
    "current_speed_ms",
    "current_direction_deg",
    "sst_deviation",
    "current_deviation",
    "ocean_features_available",
    "ocean_data_completeness"
]

print("Feature engineering completed!")
print("\nNew dataset shape:", ocean_df.shape)

print("\nEngineered features:")
print(ocean_df[engineered_features].head(10))
```

```python
# ---------------------------------------------------
# HSI MODEL - DATA-DRIVEN REFERENCE PARAMETERS
# ---------------------------------------------------

# Chlorophyll reference range
chl_q10 = ocean_df["chlorophyll_mean"].quantile(0.10)
chl_q50 = ocean_df["chlorophyll_mean"].quantile(0.50)
chl_q90 = ocean_df["chlorophyll_mean"].quantile(0.90)

# SST reference range
sst_q10 = ocean_df["sst_c"].quantile(0.10)
sst_q50 = ocean_df["sst_c"].quantile(0.50)
sst_q90 = ocean_df["sst_c"].quantile(0.90)

# Current reference range
cur_q10 = ocean_df["current_speed_ms"].quantile(0.10)
cur_q50 = ocean_df["current_speed_ms"].quantile(0.50)
cur_q90 = ocean_df["current_speed_ms"].quantile(0.90)


print("HSI reference parameters")
print("=" * 40)

print("\nChlorophyll:")
print(f"10th percentile : {chl_q10:.4f}")
print(f"Median          : {chl_q50:.4f}")
print(f"90th percentile : {chl_q90:.4f}")

print("\nSST:")
print(f"10th percentile : {sst_q10:.4f}")
print(f"Median          : {sst_q50:.4f}")
print(f"90th percentile : {sst_q90:.4f}")

print("\nCurrent speed:")
print(f"10th percentile : {cur_q10:.4f}")
print(f"Median          : {cur_q50:.4f}")
print(f"90th percentile : {cur_q90:.4f}")
```

```python
# ============================================================
# STEP 13: HSI / FISHING SUITABILITY MODEL v1
# ============================================================

import numpy as np

# ------------------------------------------------------------
# Robust reference values from the dataset
# ------------------------------------------------------------

chl_low = ocean_df["chlorophyll_mean"].quantile(0.10)
chl_high = ocean_df["chlorophyll_mean"].quantile(0.90)

sst_low = ocean_df["sst_c"].quantile(0.10)
sst_mid = ocean_df["sst_c"].quantile(0.50)
sst_high = ocean_df["sst_c"].quantile(0.90)

cur_low = ocean_df["current_speed_ms"].quantile(0.10)
cur_mid = ocean_df["current_speed_ms"].quantile(0.50)
cur_high = ocean_df["current_speed_ms"].quantile(0.90)


# ------------------------------------------------------------
# 1. Chlorophyll suitability
# Higher chlorophyll -> higher suitability
# Values above 90th percentile are capped at 1
# ------------------------------------------------------------

def chlorophyll_hsi(x):
    if pd.isna(x):
        return np.nan

    score = (x - chl_low) / (chl_high - chl_low)
    return np.clip(score, 0, 1)


# ------------------------------------------------------------
# 2. SST suitability
# Best suitability around dataset median.
# Score decreases as SST moves away from the median.
# ------------------------------------------------------------

def sst_hsi(x):
    if pd.isna(x):
        return np.nan

    if x <= sst_mid:
        score = (x - sst_low) / (sst_mid - sst_low)
    else:
        score = (sst_high - x) / (sst_high - sst_mid)

    return np.clip(score, 0, 1)


# ------------------------------------------------------------
# 3. Current suitability
# Best suitability around median current speed.
# ------------------------------------------------------------

def current_hsi(x):
    if pd.isna(x):
        return np.nan

    if x <= cur_mid:
        score = (x - cur_low) / (cur_mid - cur_low)
    else:
        score = (cur_high - x) / (cur_high - cur_mid)

    return np.clip(score, 0, 1)


# ------------------------------------------------------------
# Apply component HSI scores
# ------------------------------------------------------------

ocean_df["chl_hsi"] = ocean_df["chlorophyll_mean"].apply(
    chlorophyll_hsi
)

ocean_df["sst_hsi"] = ocean_df["sst_c"].apply(
    sst_hsi
)

ocean_df["current_hsi"] = ocean_df["current_speed_ms"].apply(
    current_hsi
)


# ------------------------------------------------------------
# 4. Combine available components
#
# Weights:
# Chlorophyll = 40%
# SST         = 30%
# Current     = 30%
#
# If one feature is missing, remaining available weights
# are re-normalized instead of treating missing as zero.
# ------------------------------------------------------------

weights = {
    "chl_hsi": 0.40,
    "sst_hsi": 0.30,
    "current_hsi": 0.30
}

def calculate_fishing_score(row):

    available = []

    for feature, weight in weights.items():
        if pd.notna(row[feature]):
            available.append((row[feature], weight))

    if len(available) == 0:
        return np.nan

    total_weight = sum(weight for score, weight in available)

    score = sum(
        score * weight
        for score, weight in available
    ) / total_weight

    return np.clip(score, 0, 1)


ocean_df["fishing_score"] = ocean_df.apply(
    calculate_fishing_score,
    axis=1
)


# ------------------------------------------------------------
# 5. Confidence based on data completeness
# ------------------------------------------------------------

ocean_df["confidence"] = ocean_df["ocean_data_completeness"]


# ------------------------------------------------------------
# 6. Inspect result
# ------------------------------------------------------------

result_cols = [
    "zone_id",
    "region",
    "chl_hsi",
    "sst_hsi",
    "current_hsi",
    "fishing_score",
    "confidence"
]

print("HSI model completed!")
print("\nSample results:\n")

print(
    ocean_df[result_cols]
    .head(15)
    .to_string(index=False)
)
```

```python
# ============================================================
# STEP 19: TOP FISHING ZONES -> JSON
# ============================================================

def rank_fishing_zones(region=None, top_k=5):
    """
    Rank fishing zones using confidence-adjusted fishing score.
    Optionally filter by region.
    """

    data = ocean_df.copy()

    # Region filter
    if region is not None:
        data = data[
            data["region"].astype(str).str.upper()
            == str(region).upper()
        ]

    # Only zones with a valid score
    data = data.dropna(
        subset=["adjusted_fishing_score"]
    )

    # Ranking
    data = data.sort_values(
        "adjusted_fishing_score",
        ascending=False
    ).head(top_k)

    results = []

    for rank, (_, row) in enumerate(data.iterrows(), start=1):

        results.append({
            "rank": rank,
            "zone_id": row["zone_id"],
            "region": row["region"],
            "location": {
                "latitude": float(row["latitude"]),
                "longitude": float(row["longitude"])
            },
            "fishing_score": round(
                float(row["fishing_score"]), 4
            ),
            "confidence": round(
                float(row["confidence"]), 2
            ),
            "adjusted_fishing_score": round(
                float(row["adjusted_fishing_score"]), 4
            )
        })

    return {
        "agent": "ocean_agent",
        "status": "success",
        "query": {
            "type": "top_fishing_zones",
            "region": region,
            "top_k": top_k
        },
        "results": results
    }


# ------------------------------------------------------------
# TEST: TOP 5 KERALA ZONES
# ------------------------------------------------------------

kerala_result = rank_fishing_zones(
    region="KERALA",
    top_k=5
)

print(
    json.dumps(
        kerala_result,
        indent=2
    )
)
```

```python
# ============================================================
# STEP 20: REGION-LEVEL OCEAN SUMMARY
# ============================================================

def get_region_ocean_summary(region):
    """
    Return overall Ocean Agent summary for a region.
    """

    region_data = ocean_df[
        ocean_df["region"].astype(str).str.upper()
        == str(region).upper()
    ].copy()

    if region_data.empty:
        return {
            "agent": "ocean_agent",
            "status": "error",
            "message": f"Region {region} not found."
        }

    # Valid scored zones
    scored = region_data.dropna(
        subset=["adjusted_fishing_score"]
    )

    # Best zone
    best_zone = None

    if not scored.empty:
        best = scored.sort_values(
            "adjusted_fishing_score",
            ascending=False
        ).iloc[0]

        best_zone = {
            "zone_id": best["zone_id"],
            "fishing_score": round(
                float(best["fishing_score"]), 4
            ),
            "confidence": round(
                float(best["confidence"]), 2
            )
        }

    return {
        "agent": "ocean_agent",
        "status": "success",
        "region": region.upper(),

        "zone_count": int(len(region_data)),

        "ocean_conditions": {
            "avg_sst_c": (
                None
                if region_data["sst_c"].dropna().empty
                else round(
                    float(region_data["sst_c"].mean()), 4
                )
            ),

            "avg_chlorophyll_mean": (
                None
                if region_data["chlorophyll_mean"].dropna().empty
                else round(
                    float(region_data["chlorophyll_mean"].mean()), 4
                )
            ),

            "avg_current_speed_ms": (
                None
                if region_data["current_speed_ms"].dropna().empty
                else round(
                    float(region_data["current_speed_ms"].mean()), 4
                )
            )
        },

        "fishing_score": (
            None
            if scored.empty
            else round(
                float(scored["fishing_score"].mean()), 4
            )
        ),

        "confidence": (
            None
            if region_data.empty
            else round(
                float(region_data["ocean_data_completeness"].mean()), 2
            )
        ),

        "best_zone": best_zone
    }


# ------------------------------------------------------------
# TEST
# ------------------------------------------------------------

kerala_summary = get_region_ocean_summary("KERALA")

print(
    json.dumps(
        kerala_summary,
        indent=2
    )
)
```

```python
# ============================================================
# STEP 21: COMPARE TWO FISHING ZONES
# ============================================================

def compare_ocean_conditions(zone1, zone2):
    """
    Compare two zones using ocean conditions and fishing score.
    """

    result1 = get_ocean_conditions(zone1)
    result2 = get_ocean_conditions(zone2)

    # Handle invalid zones
    if result1["status"] == "error":
        return result1

    if result2["status"] == "error":
        return result2

    score1 = result1["fishing_score"]
    score2 = result2["fishing_score"]

    if score1 is None and score2 is None:
        better_zone = None
    elif score1 is None:
        better_zone = result2["zone_id"]
    elif score2 is None:
        better_zone = result1["zone_id"]
    else:
        better_zone = (
            result1["zone_id"]
            if result1["adjusted_fishing_score"]
            >= result2["adjusted_fishing_score"]
            else result2["zone_id"]
        )

    return {
        "agent": "ocean_agent",
        "status": "success",

        "comparison": {
            "zone_1": result1,
            "zone_2": result2,
            "better_zone": better_zone
        },

        "comparison_basis": [
            "fishing_score",
            "confidence",
            "SST",
            "chlorophyll",
            "current_speed"
        ]
    }


# ------------------------------------------------------------
# TEST
# ------------------------------------------------------------

comparison_result = compare_ocean_conditions(
    "PFZ0074",
    "PFZ0075"
)

print(
    json.dumps(
        comparison_result,
        indent=2
    )
)
```

```python
# ============================================================
# STEP 22: OCEAN AGENT QUERY INTERFACE
# ============================================================

def ocean_agent_query(
    query_type,
    zone_id=None,
    region=None,
    zone1=None,
    zone2=None,
    top_k=5
):
    """
    Common interface for the Ocean Agent.

    Supported query types:
    - zone
    - region
    - rank
    - compare
    """

    query_type = str(query_type).lower().strip()

    # --------------------------------------------------------
    # 1. Single zone
    # --------------------------------------------------------
    if query_type == "zone":

        if zone_id is None:
            return {
                "agent": "ocean_agent",
                "status": "error",
                "message": "zone_id is required."
            }

        return get_ocean_conditions(zone_id)


    # --------------------------------------------------------
    # 2. Region summary
    # --------------------------------------------------------
    elif query_type == "region":

        if region is None:
            return {
                "agent": "ocean_agent",
                "status": "error",
                "message": "region is required."
            }

        return get_region_ocean_summary(region)


    # --------------------------------------------------------
    # 3. Rank fishing zones
    # --------------------------------------------------------
    elif query_type == "rank":

        return rank_fishing_zones(
            region=region,
            top_k=top_k
        )


    # --------------------------------------------------------
    # 4. Compare zones
    # --------------------------------------------------------
    elif query_type == "compare":

        if zone1 is None or zone2 is None:
            return {
                "agent": "ocean_agent",
                "status": "error",
                "message": "zone1 and zone2 are required."
            }

        return compare_ocean_conditions(
            zone1,
            zone2
        )


    # --------------------------------------------------------
    # Invalid query
    # --------------------------------------------------------
    else:

        return {
            "agent": "ocean_agent",
            "status": "error",
            "message": (
                "Unsupported query_type. "
                "Use: zone, region, rank, compare."
            )
        }


# ============================================================
# TEST
# ============================================================

test_result = ocean_agent_query(
    query_type="zone",
    zone_id="PFZ0074"
)

print(
    json.dumps(
        test_result,
        indent=2
    )
)
```

```python
# ============================================================
# STEP 27: VALIDATE OCEAN AGENT OUTPUT
# ============================================================

def validate_ocean_response(response):
    """
    Check whether Ocean Agent response follows
    the required JSON contract.
    """

    required_keys = [
        "agent",
        "status",
        "zone_id",
        "region",
        "location",
        "ocean_conditions",
        "hsi_components",
        "fishing_score",
        "confidence",
        "data_quality",
        "adjusted_fishing_score"
    ]

    missing_keys = [
        key for key in required_keys
        if key not in response
    ]

    if missing_keys:
        return {
            "valid": False,
            "missing_keys": missing_keys
        }

    # Score validation
    score = response["fishing_score"]

    if score is not None and not (0 <= score <= 1):
        return {
            "valid": False,
            "error": "fishing_score must be between 0 and 1."
        }

    # Confidence validation
    confidence = response["confidence"]

    if confidence is not None and not (0 <= confidence <= 1):
        return {
            "valid": False,
            "error": "confidence must be between 0 and 1."
        }

    # Adjusted score validation
    adjusted = response["adjusted_fishing_score"]

    if adjusted is not None and not (0 <= adjusted <= 1):
        return {
            "valid": False,
            "error": "adjusted_fishing_score must be between 0 and 1."
        }

    return {
        "valid": True,
        "message": "Ocean Agent response follows the JSON contract."
    }


# ------------------------------------------------------------
# TEST
# ------------------------------------------------------------

sample_response = get_ocean_conditions("PFZ0074")

validation_result = validate_ocean_response(
    sample_response
)

print(
    json.dumps(
        validation_result,
        indent=2
    )
)
```

```python
# ============================================================
# STEP 33: OCEAN AGENT PARAMETER VALIDATION
# ============================================================

def run_ocean_agent(request):

    # --------------------------------------------------------
    # 1. Request must be a dictionary
    # --------------------------------------------------------
    if not isinstance(request, dict):
        return {
            "agent": "ocean_agent",
            "status": "error",
            "error": "invalid_request",
            "message": "Request must be a JSON object."
        }

    # --------------------------------------------------------
    # 2. query_type required
    # --------------------------------------------------------
    query_type = request.get("query_type")

    supported = ["zone", "region", "rank", "compare"]

    if query_type is None:
        return {
            "agent": "ocean_agent",
            "status": "error",
            "error": "missing_query_type",
            "message": "query_type is required.",
            "supported_query_types": supported
        }

    if query_type not in supported:
        return {
            "agent": "ocean_agent",
            "status": "error",
            "error": "unsupported_query_type",
            "message": f"Unsupported query type: {query_type}",
            "supported_query_types": supported
        }

    # --------------------------------------------------------
    # 3. Validate ZONE request
    # --------------------------------------------------------
    if query_type == "zone":

        zone_id = request.get("zone_id")

        if not zone_id:
            return {
                "agent": "ocean_agent",
                "status": "error",
                "error": "missing_zone_id",
                "message": "zone_id is required for zone query."
            }

    # --------------------------------------------------------
    # 4. Validate REGION request
    # --------------------------------------------------------
    if query_type == "region":

        region = request.get("region")

        if not region:
            return {
                "agent": "ocean_agent",
                "status": "error",
                "error": "missing_region",
                "message": "region is required for region query."
            }

    # --------------------------------------------------------
    # 5. Validate RANK request
    # --------------------------------------------------------
    if query_type == "rank":

        top_k = request.get("top_k", 5)

        if not isinstance(top_k, int) or isinstance(top_k, bool):
            return {
                "agent": "ocean_agent",
                "status": "error",
                "error": "invalid_top_k",
                "message": "top_k must be an integer."
            }

        if top_k < 1:
            return {
                "agent": "ocean_agent",
                "status": "error",
                "error": "invalid_top_k",
                "message": "top_k must be greater than 0."
            }

    # --------------------------------------------------------
    # 6. Validate COMPARE request
    # --------------------------------------------------------
    if query_type == "compare":

        zone1 = request.get("zone1")
        zone2 = request.get("zone2")

        if not zone1:
            return {
                "agent": "ocean_agent",
                "status": "error",
                "error": "missing_zone1",
                "message": "zone1 is required for compare query."
            }

        if not zone2:
            return {
                "agent": "ocean_agent",
                "status": "error",
                "error": "missing_zone2",
                "message": "zone2 is required for compare query."
            }

    # --------------------------------------------------------
    # 7. If validation passes → call router
    # --------------------------------------------------------
    return ocean_agent_router(
        query_type=query_type,
        zone_id=request.get("zone_id"),
        region=request.get("region"),
        zone1=request.get("zone1"),
        zone2=request.get("zone2"),
        top_k=request.get("top_k", 5)
    )


print("Parameter validation layer ready.")
```

```python
# ============================================================
# STEP 37: NATURAL LANGUAGE QUERY PARSER
# ============================================================

def parse_ocean_query(query):

    if not isinstance(query, str) or not query.strip():
        return {
            "status": "error",
            "error": "invalid_query",
            "message": "Query must be a non-empty string."
        }

    q = query.lower().strip()

    # --------------------------------------------------------
    # Detect REGION
    # --------------------------------------------------------
    regions = [
        "KERALA",
        "MAHARASHTRA",
        "NORTH TAMILNADU",
        "SOUTH ANDHRAPRADESH",
        "ANDAMAN",
        "KARNATAKA",
        "LAKSHADWEEP",
        "GOA",
        "INDIAN_MARINE"
    ]

    detected_region = None

    for region in regions:
        if region.lower() in q:
            detected_region = region
            break

    # --------------------------------------------------------
    # Detect ZONE ID
    # --------------------------------------------------------
    import re

    zone_match = re.search(
        r"\b(?:PFZ|GRID)\d+\b",
        query.upper()
    )

    detected_zone = zone_match.group(0) if zone_match else None

    # --------------------------------------------------------
    # Detect query type
    # --------------------------------------------------------

    # Compare
    if any(word in q for word in [
        "compare",
        "comparison",
        "which is better",
        "better zone"
    ]):

        zones = re.findall(
            r"\b(?:PFZ|GRID)\d+\b",
            query.upper()
        )

        if len(zones) >= 2:
            return {
                "status": "success",
                "query_type": "compare",
                "zone1": zones[0],
                "zone2": zones[1]
            }

    # Rank
    if any(word in q for word in [
        "best",
        "top",
        "highest",
        "rank",
        "good fishing zones"
    ]):

        return {
            "status": "success",
            "query_type": "rank",
            "region": detected_region,
            "top_k": 5
        }

    # Region summary
    if any(word in q for word in [
        "region summary",
        "region conditions",
        "overall conditions",
        "overall ocean",
        "average conditions"
    ]):

        return {
            "status": "success",
            "query_type": "region",
            "region": detected_region
        }

    # Zone
    if detected_zone:

        return {
            "status": "success",
            "query_type": "zone",
            "zone_id": detected_zone
        }

    # --------------------------------------------------------
    # Could not understand
    # --------------------------------------------------------

    return {
        "status": "error",
        "error": "query_not_understood",
        "message": "Could not determine the Ocean Agent query type."
    }


print("Natural language parser ready.")
```

```python
# ============================================================
# STEP 41: HUMAN-READABLE OCEAN AGENT ANSWER
# ============================================================

def generate_ocean_answer(response):

    if response.get("status") != "success":
        error = response.get("error", {})

        if isinstance(error, dict):
            message = error.get("message", "Unknown error.")
        else:
            message = response.get("message", str(error))

        return message

    # --------------------------------------------------------
    # ZONE RESPONSE
    # --------------------------------------------------------
    if "zone_id" in response and "ocean_conditions" in response:

        zone = response["zone_id"]
        region = response["region"]
        conditions = response["ocean_conditions"]

        score = response.get("fishing_score")
        confidence = response.get("confidence")

        sst = conditions.get("sst_c")
        chl = conditions.get("chlorophyll_mean")
        current = conditions.get("current_speed_ms")

        answer = (
            f"{zone} is located in {region}. "
            f"The SST is {sst:.2f}°C and chlorophyll is {chl:.3f}."
        )

        if current is not None:
            answer += f" Current speed is {current:.3f} m/s."
        else:
            answer += " Current data is unavailable."

        if score is not None:
            answer += (
                f" The environmental fishing suitability score is "
                f"{score:.2f}."
            )

        if confidence is not None:
            answer += (
                f" Data completeness/confidence is "
                f"{confidence:.0%}."
            )

        return answer

    # --------------------------------------------------------
    # REGION RESPONSE
    # --------------------------------------------------------
    if "zone_count" in response:

        region = response["region"]
        conditions = response["ocean_conditions"]

        return (
            f"{region} contains {response['zone_count']} zones. "
            f"Average SST is {conditions['avg_sst_c']:.2f}°C, "
            f"average chlorophyll is "
            f"{conditions['avg_chlorophyll_mean']:.3f}, "
            f"and average current speed is "
            f"{conditions['avg_current_speed_ms']:.3f} m/s. "
            f"The regional environmental fishing suitability score "
            f"is {response['fishing_score']:.2f}."
        )

    # --------------------------------------------------------
    # RANK RESPONSE
    # --------------------------------------------------------
    if "results" in response:

        results = response["results"]

        if not results:
            return "No suitable fishing zones were found."

        region = response["query"].get("region")

        top = results[0]

        answer = (
            f"The top fishing-suitability zone"
            f"{' in ' + region if region else ''} "
            f"is {top['zone_id']} with a score of "
            f"{top['fishing_score']:.2f}."
        )

        if len(results) > 1:
            answer += " Other top zones are: "

            others = [
                f"{r['zone_id']} ({r['fishing_score']:.2f})"
                for r in results[1:]
            ]

            answer += ", ".join(others) + "."

        return answer

    # --------------------------------------------------------
    # COMPARE RESPONSE
    # --------------------------------------------------------
    if "comparison" in response:

        comparison = response["comparison"]

        zone1 = comparison["zone_1"]
        zone2 = comparison["zone_2"]
        better = comparison["better_zone"]

        return (
            f"{better} has the higher environmental fishing "
            f"suitability compared with the other zone. "
            f"{zone1['zone_id']} score: "
            f"{zone1['fishing_score']:.2f}; "
            f"{zone2['zone_id']} score: "
            f"{zone2['fishing_score']:.2f}."
        )

    return "Ocean Agent returned a successful response."


print("Human-readable answer layer ready.")
```

```python
# ============================================================
# STEP 42: FINAL NATURAL LANGUAGE RESPONSE
# ============================================================

def ask_ocean_agent(query):

    # --------------------------------------------------------
    # 1. Parse natural-language query
    # --------------------------------------------------------
    parsed_request = parse_ocean_query(query)

    # Parser error
    if parsed_request.get("status") != "success":

        error_response = {
            "agent": "ocean_agent",
            "status": "error",
            "error": {
                "code": parsed_request.get(
                    "error",
                    "query_parse_error"
                ),
                "message": parsed_request.get(
                    "message",
                    "Could not understand the query."
                )
            }
        }

        return {
            "answer": error_response["error"]["message"],
            "data": error_response
        }

    # --------------------------------------------------------
    # 2. Remove parser-only status
    # --------------------------------------------------------
    agent_request = {
        key: value
        for key, value in parsed_request.items()
        if key != "status"
    }

    # --------------------------------------------------------
    # 3. Run Ocean Agent
    # --------------------------------------------------------
    response = run_ocean_agent(agent_request)

    # --------------------------------------------------------
    # 4. Generate human-readable answer
    # --------------------------------------------------------
    answer = generate_ocean_answer(response)

    # --------------------------------------------------------
    # 5. Return both answer + machine-readable data
    # --------------------------------------------------------
    return {
        "answer": answer,
        "data": response
    }


print("Final natural-language Ocean Agent ready.")
```

```python
class OceanAgentState(TypedDict, total=False):
    user_query: str

    query_type: Optional[str]
    zone_id: Optional[str]
    region: Optional[str]
    zone1: Optional[str]
    zone2: Optional[str]
    top_k: Optional[int]

    agent_response: Dict[str, Any]
    answer: str
    status: str


print("✅ OceanAgentState created")
```

```python
def parse_query_node(state: OceanAgentState) -> OceanAgentState:
    query = state.get("user_query", "")

    parsed = parse_ocean_query(query)

    return {
        **state,
        **parsed,
        "status": "parsed"
    }


print("✅ parse_query_node created")
```

```python
def ocean_tool_node(state: OceanAgentState) -> OceanAgentState:
    request = {
        "query_type": state.get("query_type"),
        "zone_id": state.get("zone_id"),
        "region": state.get("region"),
        "zone1": state.get("zone1"),
        "zone2": state.get("zone2"),
        "top_k": state.get("top_k", 5),
    }

    response = run_ocean_agent(request)

    return {
        **state,
        "agent_response": response,
        "status": "tool_completed"
    }


print("✅ ocean_tool_node created")
```

```python

def answer_node(state: OceanAgentState) -> OceanAgentState:
    response = state.get("agent_response", {})

    answer = generate_ocean_answer(response)

    return {
        **state,
        "answer": answer,
        "status": "completed"
    }


print("✅ answer_node created")
```

```python
builder = StateGraph(OceanAgentState)

# Nodes
builder.add_node("parse_query", parse_query_node)
builder.add_node("ocean_tool", ocean_tool_node)
builder.add_node("answer", answer_node)

# Flow
builder.add_edge(START, "parse_query")
builder.add_edge("parse_query", "ocean_tool")
builder.add_edge("ocean_tool", "answer")
builder.add_edge("answer", END)

# Compile
ocean_graph = builder.compile()

print("✅ Ocean Agent LangGraph compiled successfully!")
```

```python
def validate_ocean_agent():

    tests = [
        {
            "name": "Zone Query",
            "query": "What are the ocean conditions at PFZ0074?",
            "expected_type": "zone"
        },
        {
            "name": "Region Ranking",
            "query": "What are the best fishing zones in Kerala?",
            "expected_type": "rank"
        },
        {
            "name": "Region Summary",
            "query": "Give me the overall ocean conditions of Kerala.",
            "expected_type": "region"
        },
        {
            "name": "Zone Comparison",
            "query": "Compare PFZ0074 and PFZ0075.",
            "expected_type": "compare"
        },
        {
            "name": "Missing Data",
            "query": "What are the ocean conditions at PFZ0001?",
            "expected_type": "zone"
        }
    ]

    results = []

    for test in tests:
        try:
            result = ocean_graph.invoke({
                "user_query": test["query"]
            })

            passed = (
                result.get("status") == "completed"
                and result.get("query_type") == test["expected_type"]
                and bool(result.get("agent_response"))
            )

            results.append({
                "test": test["name"],
                "status": "PASS" if passed else "FAIL",
                "query_type": result.get("query_type"),
                "agent_status": result.get("agent_response", {}).get("status")
            })

        except Exception as e:
            results.append({
                "test": test["name"],
                "status": "FAIL",
                "query_type": None,
                "agent_status": str(e)
            })

    return results


validation_results = validate_ocean_agent()

for r in validation_results:
    print(
        f"{r['status']:4} | "
        f"{r['test']:20} | "
        f"type={r['query_type']}"
    )
```

```python
def validate_ocean_contract(response: dict) -> bool:
    """Validate the final Ocean → Decision Agent response."""

    required_fields = [
        "agent",
        "status",
        "query_type",
        "request",
        "data",
        "answer"
    ]

    # Top-level fields
    for field in required_fields:
        if field not in response:
            return False

    if response["agent"] != "ocean_agent":
        return False

    if response["status"] != "completed":
        return False

    if not isinstance(response["data"], list):
        return False

    if not isinstance(response["answer"], str):
        return False

    # Validate each result
    for item in response["data"]:
        if "zone_id" not in item:
            return False

        if "region" not in item:
            return False

        if "adjusted_fishing_score" not in item:
            return False

        if "confidence" not in item:
            return False

    return True


print("Contract valid:", validate_ocean_contract(final_output))
```

```python
def find_high_chlorophyll_favourable_sst(top_k=5):
    """
    Find regions/zones having relatively high chlorophyll
    and favourable SST based on the existing Ocean HSI framework.
    """

    data = ocean_df.copy()

    # Remove rows where either required feature is missing
    data = data.dropna(subset=["chlorophyll_mean", "sst_c"])

    # High chlorophyll: top 20% of observed chlorophyll
    chl_threshold = data["chlorophyll_mean"].quantile(0.80)

    # Favourable SST: use existing SST HSI
    data["sst_suitability"] = data["sst_c"].apply(sst_hsi)

    # Keep high-chlorophyll + favourable-SST locations
    suitable = data[
        (data["chlorophyll_mean"] >= chl_threshold) &
        (data["sst_suitability"] >= 0.70)
    ].copy()

    # Rank by chlorophyll + SST suitability
    suitable["combined_environment_score"] = (
        0.5 * suitable["sst_suitability"] +
        0.5 * suitable["chl_hsi"]
    )

    # Region-level summary
    region_summary = (
        suitable.groupby("region")
        .agg(
            zones=("zone_id", "count"),
            avg_chlorophyll=("chlorophyll_mean", "mean"),
            avg_sst=("sst_c", "mean"),
            avg_sst_suitability=("sst_suitability", "mean"),
            avg_environment_score=("combined_environment_score", "mean")
        )
        .sort_values("avg_environment_score", ascending=False)
        .head(top_k)
        .reset_index()
    )

    return region_summary


print("✅ High chlorophyll + favourable SST function created")
```

```python
import numpy as np

def calculate_distance_km(lat1, lon1, lat2, lon2):
    """
    Calculate approximate distance between two latitude/longitude points
    using the Haversine formula.
    """
    R = 6371.0  # Earth radius in km

    lat1, lon1, lat2, lon2 = map(
        np.radians,
        [lat1, lon1, lat2, lon2]
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    )

    return 2 * R * np.arcsin(np.sqrt(a))


def find_nearest_pfz(latitude, longitude, top_k=5):
    """
    Find the nearest PFZ zones from a given latitude/longitude.
    """

    data = ocean_df.copy()

    data = data.dropna(subset=["latitude", "longitude"])

    data["distance_from_reference_km"] = calculate_distance_km(
        latitude,
        longitude,
        data["latitude"].values,
        data["longitude"].values
    )

    result = (
        data.sort_values("distance_from_reference_km")
        .head(top_k)
        [
            [
                "zone_id",
                "region",
                "latitude",
                "longitude",
                "distance_from_reference_km",
                "adjusted_fishing_score"
            ]
        ]
        .copy()
    )

    return result.reset_index(drop=True)


print("✅ Nearest PFZ function created")
```

```python
def analyze_ocean_productivity(region):
    """
    Analyze ocean conditions that may indicate lower
    environmental fishing suitability in a region.

    Note:
    This does NOT measure actual fish productivity decline.
    It evaluates current environmental indicators only.
    """

    data = ocean_df[
        ocean_df["region"].str.upper() == region.upper()
    ].copy()

    if data.empty:
        return {
            "region": region.upper(),
            "status": "error",
            "message": "Region not found."
        }

    summary = {
        "region": region.upper(),
        "zones_analyzed": len(data),
        "avg_chlorophyll": data["chlorophyll_mean"].mean(),
        "avg_sst": data["sst_c"].mean(),
        "avg_current_speed": data["current_speed_ms"].mean(),
        "avg_chl_hsi": data["chl_hsi"].mean(),
        "avg_sst_hsi": data["sst_hsi"].mean(),
        "avg_current_hsi": data["current_hsi"].mean(),
        "avg_fishing_suitability": data["fishing_score"].mean(),
        "avg_confidence": data["confidence"].mean()
    }

    factors = []

    if summary["avg_chl_hsi"] < 0.5:
        factors.append("relatively low chlorophyll suitability")

    if summary["avg_sst_hsi"] < 0.5:
        factors.append("less favourable SST conditions")

    if summary["avg_current_hsi"] < 0.5:
        factors.append("less favourable current conditions")

    if not factors:
        factors.append("no strongly unfavourable ocean indicator detected")

    summary["possible_environmental_factors"] = factors

    return summary


print("✅ Ocean productivity analysis function created")
```

```python
def find_highest_pfz_confidence(top_k=5):
    data = ocean_df.copy()

    data = data.dropna(
        subset=["zone_id", "region", "confidence"]
    ).copy()

    result = (
        data.sort_values(
            ["confidence", "adjusted_fishing_score"],
            ascending=[False, False]
        )
        .head(top_k)
        .reset_index(drop=True)
    )

    result.insert(0, "rank", range(1, len(result) + 1))

    output_cols = [
        "rank",
        "zone_id",
        "region",
        "confidence",
        "fishing_score",
        "adjusted_fishing_score"
    ]

    return result[output_cols]


print("✅ Highest PFZ confidence function ready")
```

```python
def parse_ocean_query(query: str) -> dict:
    query = query.lower().strip()

    regions = [
        "KERALA",
        "MAHARASHTRA",
        "NORTH TAMILNADU",
        "SOUTH ANDHRAPRADESH",
        "ANDAMAN",
        "KARNATAKA",
        "LAKSHADWEEP",
        "GOA",
        "INDIAN_MARINE"
    ]

    detected_region = None

    for region in regions:
        region_text = region.lower().replace("_", " ")
        if region_text in query:
            detected_region = region
            break

    # Highest PFZ confidence
    if (
        "highest" in query
        and "confidence" in query
        and ("pfz" in query or "zone" in query)
    ):
        return {
            "query_type": "highest_confidence",
            "zone_id": None,
            "region": detected_region,
            "zone1": None,
            "zone2": None,
            "top_k": 5
        }

    # Nearest PFZ
    if (
        "nearest" in query
        and ("pfz" in query or "potential fishing zone" in query)
    ):
        if "kochi" in query:
            return {
                "query_type": "nearest_pfz",
                "zone_id": None,
                "region": None,
                "zone1": None,
                "zone2": None,
                "top_k": 5
            }

    # High chlorophyll + favourable SST
    if (
        "chlorophyll" in query
        and ("sst" in query or "sea surface temperature" in query)
    ):
        return {
            "query_type": "high_chlorophyll_sst",
            "zone_id": None,
            "region": detected_region,
            "zone1": None,
            "zone2": None,
            "top_k": 5
        }

    # Productivity analysis
    if (
        ("productivity" in query or "fish productivity" in query)
        and ("declined" in query or "decline" in query or "decreased" in query)
    ):
        return {
            "query_type": "productivity_analysis",
            "zone_id": None,
            "region": detected_region,
            "zone1": None,
            "zone2": None,
            "top_k": 5
        }

    # Zone ID detection
    import re

    zone_matches = re.findall(
        r"\b(?:PFZ|GRID)\d+\b",
        query.upper()
    )

    # Comparison
    if (
        "compare" in query
        or "comparison" in query
        or "versus" in query
        or " vs " in query
    ):
        if len(zone_matches) >= 2:
            return {
                "query_type": "compare",
                "zone_id": None,
                "region": None,
                "zone1": zone_matches[0],
                "zone2": zone_matches[1],
                "top_k": 5
            }

    # Ranking
    if (
        "best" in query
        or "top" in query
        or "highest" in query
        or "rank" in query
    ):
        return {
            "query_type": "rank",
            "zone_id": None,
            "region": detected_region,
            "zone1": None,
            "zone2": None,
            "top_k": 5
        }

    # Region summary
    if (
        "overall" in query
        or "summary" in query
        or "conditions of" in query
        or "conditions in" in query
    ):
        return {
            "query_type": "region_summary",
            "zone_id": None,
            "region": detected_region,
            "zone1": None,
            "zone2": None,
            "top_k": 5
        }

    # Zone query
    if zone_matches:
        return {
            "query_type": "zone",
            "zone_id": zone_matches[0],
            "region": None,
            "zone1": None,
            "zone2": None,
            "top_k": 5
        }

    return {
        "query_type": "unknown",
        "zone_id": None,
        "region": detected_region,
        "zone1": None,
        "zone2": None,
        "top_k": 5
    }


print("✅ Parser updated with highest-confidence query")
```

```python
def ocean_tool_node(state: OceanAgentState) -> OceanAgentState:
    query_type = state.get("query_type")

    # Highest PFZ confidence
    if query_type == "highest_confidence":
        results = find_highest_pfz_confidence(
            state.get("top_k", 5)
        )

        agent_response = {
            "agent": "ocean_agent",
            "status": "success",
            "query": {
                "type": "highest_confidence",
                "region": state.get("region"),
                "top_k": state.get("top_k", 5)
            },
            "results": results.to_dict(orient="records")
        }

    # Productivity analysis
    elif query_type == "productivity_analysis":
        region = state.get("region")

        if not region:
            agent_response = {
                "agent": "ocean_agent",
                "status": "error",
                "results": [],
                "message": "Please specify a coastal region."
            }
        else:
            analysis = analyze_ocean_productivity(region)

            agent_response = {
                "agent": "ocean_agent",
                "status": "success",
                "query": {
                    "type": "productivity_analysis",
                    "region": region
                },
                "results": [analysis]
            }

    # Nearest PFZ
    elif query_type == "nearest_pfz":
        nearest = find_nearest_pfz(
            latitude=9.9312,
            longitude=76.2673,
            top_k=state.get("top_k", 5)
        )

        agent_response = {
            "agent": "ocean_agent",
            "status": "success",
            "query": {
                "type": "nearest_pfz",
                "reference_location": "Kochi",
                "latitude": 9.9312,
                "longitude": 76.2673,
                "top_k": state.get("top_k", 5)
            },
            "results": nearest.to_dict(orient="records")
        }

    # High chlorophyll + favourable SST
    elif query_type == "high_chlorophyll_sst":
        results = find_high_chlorophyll_favourable_sst(
            state.get("top_k", 5)
        )

        agent_response = {
            "agent": "ocean_agent",
            "status": "success",
            "query": {
                "type": "high_chlorophyll_sst",
                "top_k": state.get("top_k", 5)
            },
            "results": results.to_dict(orient="records")
        }

    # Existing queries
    else:
        request = {
            "query_type": query_type,
            "zone_id": state.get("zone_id"),
            "region": state.get("region"),
            "zone1": state.get("zone1"),
            "zone2": state.get("zone2"),
            "top_k": state.get("top_k", 5)
        }

        agent_response = run_ocean_agent(request)

    return {
        **state,
        "agent_response": agent_response,
        "status": "tool_completed"
    }


print("✅ Tool node updated")
```

```python
def answer_node(state: OceanAgentState) -> OceanAgentState:
    query_type = state.get("query_type")
    response = state.get("agent_response", {})
    results = response.get("results", [])

    if query_type == "highest_confidence":
        if not results:
            answer = "No zone confidence data is available."
        else:
            highest = results[0]

            answer = (
                f"The zone with the highest PFZ confidence is "
                f"{highest.get('zone_id')} in "
                f"{highest.get('region')} with "
                f"{highest.get('confidence', 0):.0%} confidence."
            )

            if len(results) > 1:
                others = ", ".join(
                    f"{r.get('zone_id')} ({r.get('confidence', 0):.0%})"
                    for r in results[1:]
                )
                answer += f" Other high-confidence zones are: {others}."

    elif query_type == "productivity_analysis":
        if not results:
            answer = "No ocean productivity analysis is available."
        else:
            data = results[0]

            factors = data.get(
                "possible_environmental_factors",
                []
            )

            answer = (
                f"For {data.get('region')}, the current ocean conditions show "
                f"average chlorophyll of "
                f"{data.get('avg_chlorophyll', 0):.3f}, "
                f"average SST of "
                f"{data.get('avg_sst', 0):.2f}°C, and "
                f"average current speed of "
                f"{data.get('avg_current_speed', 0):.3f} m/s. "
                f"The environmental fishing suitability score is "
                f"{data.get('avg_fishing_suitability', 0):.2f}. "
                f"Data confidence is "
                f"{data.get('avg_confidence', 0):.0%}. "
                f"Possible environmental factor(s): "
                f"{', '.join(factors)}. "
                f"This analysis evaluates environmental conditions only; "
                f"it does not establish an actual historical decline "
                f"in fish productivity."
            )

    elif query_type == "nearest_pfz":
        if not results:
            answer = "No PFZ was found near the reference location."
        else:
            nearest = results[0]

            answer = (
                f"The nearest PFZ from Kochi is "
                f"{nearest.get('zone_id')} in "
                f"{nearest.get('region')}, approximately "
                f"{nearest.get('distance_from_reference_km', 0):.2f} km away."
            )

    elif query_type == "high_chlorophyll_sst":
        if not results:
            answer = "No suitable regions were found."
        else:
            parts = []

            for r in results:
                parts.append(
                    f"{r.get('region')} "
                    f"({r.get('zones')} zones, "
                    f"chlorophyll {r.get('avg_chlorophyll', 0):.3f}, "
                    f"SST {r.get('avg_sst', 0):.2f}°C)"
                )

            answer = (
                "Regions with high chlorophyll and favourable SST are: "
                + ", ".join(parts) + "."
            )

    else:
        answer = generate_ocean_answer(response)

    return {
        **state,
        "answer": answer,
        "status": "completed"
    }


print("✅ Answer node updated")
```
