```python
# ============================================================
# M3 ROUTE AGENT v2
# STEP 1: LOAD ROUTING DATASET
# ============================================================

import pandas as pd
import numpy as np
import json

FILE_PATH = "unified_routing 2.csv"

route_df = pd.read_csv(FILE_PATH)

print("Route dataset loaded successfully!")
print("=" * 60)

print("Dataset shape:", route_df.shape)

print("\nColumns:")
for i, col in enumerate(route_df.columns, start=1):
    print(f"{i}. {col}")

print("\nFirst 5 rows:")
display(route_df.head())
```

```python
# ============================================================
# STEP 2: ROUTE FEATURE AVAILABILITY / MISSING DATA
# ============================================================

print("ROUTE FEATURE AVAILABILITY")
print("=" * 60)

# Routing variables used by M3
route_cols = [
    "distance_km",
    "travel_time_hr",
    "fuel_l_proxy",
    "water_depth_m",
    "current_u_ms",
    "current_v_ms",
    "current_speed_ms",
    "geofence_caution",
    "inside_mpa",
    "mpa_distance_km",
    "mpa_caution",
    "eez_boundary_distance_km"
]

# Keep only columns actually present
route_cols = [col for col in route_cols if col in route_df.columns]

# Missing values
print("\nMissing values in routing features:\n")
print(route_df[route_cols].isnull().sum())

# Availability
print("\nFeature availability:\n")

for col in route_cols:
    available = route_df[col].notna().sum()
    total = len(route_df)
    percentage = (available / total) * 100

    print(
        f"{col}: "
        f"{available}/{total} available "
        f"({percentage:.2f}%)"
    )

# Statistics
print("\nRouting feature statistics:\n")

display(
    route_df[route_cols]
    .describe()
    .T
)
```

```python
# ============================================================
# STEP 3: ROUTE FEATURE ENGINEERING
# ============================================================

# Keep original dataset unchanged
route_features_df = route_df.copy()

# ------------------------------------------------------------
# 1. Current direction from U and V components
# ------------------------------------------------------------

route_features_df["current_direction_deg"] = (
    np.degrees(
        np.arctan2(
            route_features_df["current_v_ms"],
            route_features_df["current_u_ms"]
        )
    ) + 360
) % 360


# ------------------------------------------------------------
# 2. Log transformation for fuel
# Helps reduce the influence of very large fuel values
# ------------------------------------------------------------

route_features_df["fuel_log"] = np.log1p(
    route_features_df["fuel_l_proxy"]
)


# ------------------------------------------------------------
# 3. Log transformation for travel time
# ------------------------------------------------------------

route_features_df["travel_time_log"] = np.log1p(
    route_features_df["travel_time_hr"]
)


# ------------------------------------------------------------
# 4. Route-data availability
# ------------------------------------------------------------

core_route_features = [
    "distance_km",
    "travel_time_hr",
    "fuel_l_proxy",
    "water_depth_m",
    "current_speed_ms"
]

route_features_df["route_features_available"] = (
    route_features_df[core_route_features]
    .notna()
    .sum(axis=1)
)

route_features_df["route_data_completeness"] = (
    route_features_df["route_features_available"]
    / len(core_route_features)
)


# ------------------------------------------------------------
# 5. Safety / constraint availability
# ------------------------------------------------------------

safety_features = [
    "water_depth_m",
    "geofence_caution",
    "inside_mpa",
    "mpa_distance_km",
    "mpa_caution",
    "eez_boundary_distance_km"
]

available_safety_features = [
    col for col in safety_features
    if col in route_features_df.columns
]

route_features_df["safety_features_available"] = (
    route_features_df[available_safety_features]
    .notna()
    .sum(axis=1)
)

route_features_df["safety_data_completeness"] = (
    route_features_df["safety_features_available"]
    / len(available_safety_features)
)


# ------------------------------------------------------------
# 6. Inspect engineered features
# ------------------------------------------------------------

engineered_route_features = [
    "zone_id",
    "region",
    "latitude",
    "longitude",
    "distance_km",
    "travel_time_hr",
    "fuel_l_proxy",
    "fuel_log",
    "travel_time_log",
    "water_depth_m",
    "current_u_ms",
    "current_v_ms",
    "current_speed_ms",
    "current_direction_deg",
    "route_features_available",
    "route_data_completeness",
    "safety_features_available",
    "safety_data_completeness"
]

print("✅ Route feature engineering completed!")
print("\nNew dataset shape:", route_features_df.shape)

print("\nEngineered route features:")
display(
    route_features_df[engineered_route_features].head(10)
)
```

```python
# ============================================================
# STEP 4: DATA-DRIVEN ROUTE REFERENCE PARAMETERS
# ============================================================

print("ROUTE REFERENCE PARAMETERS")
print("=" * 60)

# ------------------------------------------------------------
# Distance references
# Lower distance is better
# ------------------------------------------------------------

dist_q10 = route_features_df["distance_km"].quantile(0.10)
dist_q50 = route_features_df["distance_km"].quantile(0.50)
dist_q90 = route_features_df["distance_km"].quantile(0.90)


# ------------------------------------------------------------
# Travel-time references
# Lower travel time is better
# ------------------------------------------------------------

time_q10 = route_features_df["travel_time_hr"].quantile(0.10)
time_q50 = route_features_df["travel_time_hr"].quantile(0.50)
time_q90 = route_features_df["travel_time_hr"].quantile(0.90)


# ------------------------------------------------------------
# Fuel references
# Lower fuel requirement is better
# ------------------------------------------------------------

fuel_q10 = route_features_df["fuel_l_proxy"].quantile(0.10)
fuel_q50 = route_features_df["fuel_l_proxy"].quantile(0.50)
fuel_q90 = route_features_df["fuel_l_proxy"].quantile(0.90)


# ------------------------------------------------------------
# Water-depth references
# Kept as data-driven reference values for safety scoring
# ------------------------------------------------------------

depth_q10 = route_features_df["water_depth_m"].quantile(0.10)
depth_q50 = route_features_df["water_depth_m"].quantile(0.50)
depth_q90 = route_features_df["water_depth_m"].quantile(0.90)


# ------------------------------------------------------------
# MPA-distance references
# Larger distance from MPA is potentially safer
# ------------------------------------------------------------

mpa_q10 = route_features_df["mpa_distance_km"].quantile(0.10)
mpa_q50 = route_features_df["mpa_distance_km"].quantile(0.50)
mpa_q90 = route_features_df["mpa_distance_km"].quantile(0.90)


# ------------------------------------------------------------
# Current-speed references
# Kept for route-safety/current-related analysis
# ------------------------------------------------------------

current_q10 = route_features_df["current_speed_ms"].quantile(0.10)
current_q50 = route_features_df["current_speed_ms"].quantile(0.50)
current_q90 = route_features_df["current_speed_ms"].quantile(0.90)


# ------------------------------------------------------------
# Display reference parameters
# ------------------------------------------------------------

print("\nDistance:")
print(f"10th percentile : {dist_q10:.4f}")
print(f"Median          : {dist_q50:.4f}")
print(f"90th percentile : {dist_q90:.4f}")

print("\nTravel Time:")
print(f"10th percentile : {time_q10:.4f}")
print(f"Median          : {time_q50:.4f}")
print(f"90th percentile : {time_q90:.4f}")

print("\nFuel:")
print(f"10th percentile : {fuel_q10:.4f}")
print(f"Median          : {fuel_q50:.4f}")
print(f"90th percentile : {fuel_q90:.4f}")

print("\nWater Depth:")
print(f"10th percentile : {depth_q10:.4f}")
print(f"Median          : {depth_q50:.4f}")
print(f"90th percentile : {depth_q90:.4f}")

print("\nMPA Distance:")
print(f"10th percentile : {mpa_q10:.4f}")
print(f"Median          : {mpa_q50:.4f}")
print(f"90th percentile : {mpa_q90:.4f}")

print("\nCurrent Speed:")
print(f"10th percentile : {current_q10:.4f}")
print(f"Median          : {current_q50:.4f}")
print(f"90th percentile : {current_q90:.4f}")
```

```python
# ============================================================
# STEP 5: DISTANCE SUITABILITY SCORE
# ============================================================

def distance_route_score(x):
    """
    Convert route distance into a 0-1 suitability score.

    Lower distance = better route suitability.
    """

    if pd.isna(x):
        return np.nan

    # Very short distances = maximum suitability
    if x <= dist_q10:
        return 1.0

    # Very long distances = minimum suitability
    if x >= dist_q90:
        return 0.0

    # Linear scaling between q10 and q90
    score = (
        (dist_q90 - x)
        / (dist_q90 - dist_q10)
    )

    return np.clip(score, 0, 1)


# Apply distance scoring
route_features_df["distance_score"] = (
    route_features_df["distance_km"]
    .apply(distance_route_score)
)


# ------------------------------------------------------------
# Inspect results
# ------------------------------------------------------------

print("✅ Distance suitability score calculated")

print("\nDistance score statistics:")
print(
    route_features_df["distance_score"]
    .describe()
)

print("\nSample distance scores:")

display(
    route_features_df[
        [
            "zone_id",
            "distance_km",
            "distance_score"
        ]
    ].head(15)
)
```

```python
# ============================================================
# STEP 6: TRAVEL-TIME SUITABILITY SCORE
# ============================================================

def travel_time_route_score(x):
    """
    Convert travel time into a 0-1 suitability score.

    Lower travel time = better route suitability.
    """

    if pd.isna(x):
        return np.nan

    # Very short travel time = maximum suitability
    if x <= time_q10:
        return 1.0

    # Very long travel time = minimum suitability
    if x >= time_q90:
        return 0.0

    # Linear inverse scaling between q10 and q90
    score = (
        (time_q90 - x)
        / (time_q90 - time_q10)
    )

    return np.clip(score, 0, 1)


# Apply travel-time scoring
route_features_df["travel_time_score"] = (
    route_features_df["travel_time_hr"]
    .apply(travel_time_route_score)
)


# ------------------------------------------------------------
# Inspect results
# ------------------------------------------------------------

print("✅ Travel-time suitability score calculated")

print("\nTravel-time score statistics:")
print(
    route_features_df["travel_time_score"]
    .describe()
)

print("\nSample travel-time scores:")

display(
    route_features_df[
        [
            "zone_id",
            "travel_time_hr",
            "travel_time_score"
        ]
    ].head(15)
)
```

```python
# ============================================================
# STEP 7: FUEL SUITABILITY SCORE
# ============================================================

def fuel_route_score(x):
    """
    Convert fuel requirement into a 0-1 suitability score.

    Lower fuel requirement = better route suitability.
    """

    if pd.isna(x):
        return np.nan

    # Very low fuel requirement = maximum suitability
    if x <= fuel_q10:
        return 1.0

    # Very high fuel requirement = minimum suitability
    if x >= fuel_q90:
        return 0.0

    # Linear inverse scaling between q10 and q90
    score = (
        (fuel_q90 - x)
        / (fuel_q90 - fuel_q10)
    )

    return np.clip(score, 0, 1)


# Apply fuel scoring
route_features_df["fuel_score"] = (
    route_features_df["fuel_l_proxy"]
    .apply(fuel_route_score)
)


# ------------------------------------------------------------
# Inspect results
# ------------------------------------------------------------

print("✅ Fuel suitability score calculated")

print("\nFuel score statistics:")
print(
    route_features_df["fuel_score"]
    .describe()
)

print("\nSample fuel scores:")

display(
    route_features_df[
        [
            "zone_id",
            "fuel_l_proxy",
            "fuel_score"
        ]
    ].head(15)
)
```

```python
# ============================================================
# STEP 8: ROUTE EFFICIENCY SCORE
# ============================================================

# Because distance, travel time and fuel are perfectly correlated
# in this dataset, they are treated as one underlying
# route-efficiency dimension.

eff_q10 = route_features_df["distance_km"].quantile(0.10)
eff_q50 = route_features_df["distance_km"].quantile(0.50)
eff_q90 = route_features_df["distance_km"].quantile(0.90)


def route_efficiency_score(x):
    """
    Convert route distance into a normalized 0-1
    route-efficiency score.

    Lower route distance = higher efficiency.
    """

    if pd.isna(x):
        return np.nan

    if x <= eff_q10:
        return 1.0

    if x >= eff_q90:
        return 0.0

    score = (
        (eff_q90 - x)
        / (eff_q90 - eff_q10)
    )

    return np.clip(score, 0, 1)


route_features_df["efficiency_score"] = (
    route_features_df["distance_km"]
    .apply(route_efficiency_score)
)


print("✅ Route efficiency score calculated")

print("\nEfficiency reference values:")
print(f"Q10    : {eff_q10:.4f}")
print(f"Median : {eff_q50:.4f}")
print(f"Q90    : {eff_q90:.4f}")

print("\nEfficiency score statistics:")
display(
    route_features_df["efficiency_score"].describe()
)

print("\nSample:")
display(
    route_features_df[
        [
            "zone_id",
            "distance_km",
            "travel_time_hr",
            "fuel_l_proxy",
            "efficiency_score"
        ]
    ].head(15)
)
```

```python
# ============================================================
# STEP 10: WATER-DEPTH ROUTE SAFETY SCORE
# ============================================================

def depth_safety_score(x):
    """
    Convert water depth into a 0-1 route-safety score.

    Very shallow water = lower safety.
    Adequate water depth = higher safety.

    Dataset-derived references:
    q10  -> lower-depth reference
    q50  -> adequate-depth benchmark
    """

    if pd.isna(x):
        return np.nan

    # At or below the lower reference:
    # treat as lowest depth suitability
    if x <= depth_q10:
        return 0.0

    # At or above the median:
    # treat as adequate depth for this dataset
    if x >= depth_q50:
        return 1.0

    # Linear increase between q10 and median
    score = (
        (x - depth_q10)
        / (depth_q50 - depth_q10)
    )

    return np.clip(score, 0, 1)


# Apply depth safety scoring
route_features_df["depth_safety_score"] = (
    route_features_df["water_depth_m"]
    .apply(depth_safety_score)
)


# ------------------------------------------------------------
# Inspect results
# ------------------------------------------------------------

print("✅ Water-depth safety score calculated")

print("\nDepth reference values:")
print(f"Q10    : {depth_q10:.4f}")
print(f"Median : {depth_q50:.4f}")
print(f"Q90    : {depth_q90:.4f}")

print("\nDepth safety-score statistics:")
display(
    route_features_df["depth_safety_score"].describe()
)

print("\nSample:")
display(
    route_features_df[
        [
            "zone_id",
            "water_depth_m",
            "depth_safety_score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# STEP 12: ROUTE CONSTRAINT SAFETY SCORE
# ============================================================

def constraint_safety_score(row):
    """
    Calculate a 0-1 route constraint safety score.

    Each available constraint check contributes equally.

    1 = no constraint warning
    0 = constraint warning / restricted condition
    """

    checks = []

    # --------------------------------------------------------
    # MPA occupancy
    # --------------------------------------------------------
    if pd.notna(row["inside_mpa"]):
        checks.append(
            0.0 if int(row["inside_mpa"]) == 1 else 1.0
        )

    # --------------------------------------------------------
    # MPA caution
    # --------------------------------------------------------
    if pd.notna(row["mpa_caution"]):
        checks.append(
            0.0 if int(row["mpa_caution"]) == 1 else 1.0
        )

    # --------------------------------------------------------
    # Geofence caution
    # --------------------------------------------------------
    if pd.notna(row["geofence_caution"]):
        checks.append(
            0.0 if int(row["geofence_caution"]) == 1 else 1.0
        )

    # No constraint information available
    if len(checks) == 0:
        return np.nan

    return np.clip(
        sum(checks) / len(checks),
        0,
        1
    )


# Apply constraint scoring
route_features_df["constraint_safety_score"] = (
    route_features_df.apply(
        constraint_safety_score,
        axis=1
    )
)


# ------------------------------------------------------------
# Inspect results
# ------------------------------------------------------------

print("✅ Constraint safety score calculated")

print("\nConstraint safety-score distribution:")
print(
    route_features_df[
        "constraint_safety_score"
    ].value_counts(dropna=False)
)

print("\nSample:")
display(
    route_features_df[
        [
            "zone_id",
            "inside_mpa",
            "mpa_caution",
            "geofence_caution",
            "constraint_safety_score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# STEP 13: CURRENT-BASED ROUTE SAFETY SCORE
# ============================================================

# Dataset-derived current references
current_safety_q10 = route_features_df[
    "current_speed_ms"
].quantile(0.10)

current_safety_q50 = route_features_df[
    "current_speed_ms"
].quantile(0.50)

current_safety_q90 = route_features_df[
    "current_speed_ms"
].quantile(0.90)


def current_safety_score(x):
    """
    Convert current speed into a 0-1 route-safety score.

    Lower current magnitude = higher route-safety score.

    Missing current data remains missing.
    """

    if pd.isna(x):
        return np.nan

    # Low current magnitude
    if x <= current_safety_q10:
        return 1.0

    # High current magnitude
    if x >= current_safety_q90:
        return 0.0

    # Inverse linear scaling
    score = (
        (current_safety_q90 - x)
        / (
            current_safety_q90
            - current_safety_q10
        )
    )

    return np.clip(score, 0, 1)


# Apply current safety scoring
route_features_df["current_safety_score"] = (
    route_features_df["current_speed_ms"]
    .apply(current_safety_score)
)


# ------------------------------------------------------------
# Inspect results
# ------------------------------------------------------------

print("✅ Current-based route safety score calculated")

print("\nCurrent references:")
print(
    f"Q10    : {current_safety_q10:.6f}"
)
print(
    f"Median : {current_safety_q50:.6f}"
)
print(
    f"Q90    : {current_safety_q90:.6f}"
)

print("\nCurrent safety score statistics:")
display(
    route_features_df[
        "current_safety_score"
    ].describe()
)

print("\nSample:")
display(
    route_features_df[
        [
            "zone_id",
            "current_speed_ms",
            "current_safety_score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# STEP 14: COMBINED ROUTE SAFETY SCORE
# ============================================================

safety_weights = {
    "depth_safety_score": 1 / 3,
    "constraint_safety_score": 1 / 3,
    "current_safety_score": 1 / 3
}


def calculate_route_safety_score(row):
    """
    Combine independent route-safety components.

    Missing components are excluded and the remaining
    weights are re-normalized.
    """

    available = []

    for feature, weight in safety_weights.items():

        if pd.notna(row[feature]):
            available.append(
                (row[feature], weight)
            )

    # No safety information available
    if len(available) == 0:
        return np.nan

    total_weight = sum(
        weight for score, weight in available
    )

    score = sum(
        score * weight
        for score, weight in available
    ) / total_weight

    return np.clip(score, 0, 1)


# ------------------------------------------------------------
# Apply combined safety score
# ------------------------------------------------------------

route_features_df["route_safety_score"] = (
    route_features_df.apply(
        calculate_route_safety_score,
        axis=1
    )
)


# ------------------------------------------------------------
# Safety confidence
# ------------------------------------------------------------

route_features_df["route_safety_confidence"] = (
    route_features_df[
        [
            "depth_safety_score",
            "constraint_safety_score",
            "current_safety_score"
        ]
    ]
    .notna()
    .sum(axis=1)
    / 3
)


# ------------------------------------------------------------
# Inspect results
# ------------------------------------------------------------

print("✅ Combined route safety score calculated")

print("\nRoute safety score statistics:")
display(
    route_features_df[
        "route_safety_score"
    ].describe()
)

print("\nRoute safety confidence statistics:")
display(
    route_features_df[
        "route_safety_confidence"
    ].describe()
)

print("\nSample:")
display(
    route_features_df[
        [
            "zone_id",
            "depth_safety_score",
            "constraint_safety_score",
            "current_safety_score",
            "route_safety_score",
            "route_safety_confidence"
        ]
    ].head(20)
)
```

```python
# ============================================================
# STEP 15: FINAL ROUTE SCORE
# ============================================================

# ------------------------------------------------------------
# Route-score weights
# Screenshot specifies:
# Distance + Time + Fuel + Route Safety
#
# Distance/Time/Fuel are perfectly correlated in this dataset,
# so they are represented by ONE efficiency component.
#
# No numerical weights were specified in the workflow,
# therefore use neutral equal weighting:
#   Efficiency = 50%
#   Safety     = 50%
# ------------------------------------------------------------

route_score_weights = {
    "efficiency_score": 0.50,
    "route_safety_score": 0.50
}


def calculate_route_score(row):
    """
    Calculate final 0-1 Route Score.

    Missing components are excluded and the remaining
    weights are re-normalized.
    """

    available = []

    for feature, weight in route_score_weights.items():

        if pd.notna(row[feature]):
            available.append(
                (row[feature], weight)
            )

    # No usable routing information
    if len(available) == 0:
        return np.nan

    total_weight = sum(
        weight for score, weight in available
    )

    score = sum(
        score * weight
        for score, weight in available
    ) / total_weight

    return np.clip(score, 0, 1)


# ------------------------------------------------------------
# Calculate final Route Score
# ------------------------------------------------------------

route_features_df["route_score_calculated"] = (
    route_features_df.apply(
        calculate_route_score,
        axis=1
    )
)


# ------------------------------------------------------------
# Overall route confidence
# ------------------------------------------------------------

route_features_df["route_confidence"] = (
    route_features_df[
        [
            "efficiency_score",
            "route_safety_score"
        ]
    ]
    .notna()
    .sum(axis=1)
    / 2
)


# ------------------------------------------------------------
# Confidence-adjusted Route Score
# ------------------------------------------------------------

route_features_df["adjusted_route_score"] = (
    route_features_df["route_score_calculated"]
    * route_features_df["route_confidence"]
)


# ------------------------------------------------------------
# Inspect result
# ------------------------------------------------------------

print("✅ Final Route Score calculated")

print("\nRoute Score statistics:")
display(
    route_features_df[
        "route_score_calculated"
    ].describe()
)

print("\nRoute Confidence:")
display(
    route_features_df[
        "route_confidence"
    ].value_counts(dropna=False)
)

print("\nSample Route Scores:")

display(
    route_features_df[
        [
            "zone_id",
            "distance_km",
            "travel_time_hr",
            "fuel_l_proxy",
            "efficiency_score",
            "route_safety_score",
            "route_score_calculated",
            "route_confidence",
            "adjusted_route_score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# STEP 23: FINAL ROUTE AGENT JSON OUTPUT
# ============================================================

def get_route_conditions(zone_id):
    """
    Return structured JSON-compatible Route Agent output
    for a single zone.
    """

    # --------------------------------------------------------
    # Find zone
    # --------------------------------------------------------

    result = route_features_df[
        route_features_df["zone_id"]
        .astype(str)
        .str.upper()
        == str(zone_id).upper()
    ]

    # --------------------------------------------------------
    # Zone not found
    # --------------------------------------------------------

    if result.empty:
        return {
            "agent": "route_agent",
            "status": "error",
            "error": {
                "code": "zone_not_found",
                "message": f"Zone {zone_id} not found."
            }
        }

    row = result.iloc[0]

    # --------------------------------------------------------
    # Helper for safe numeric JSON values
    # --------------------------------------------------------

    def safe_float(value, decimals=4):
        if pd.isna(value):
            return None
        return round(float(value), decimals)

    # --------------------------------------------------------
    # Final structured response
    # --------------------------------------------------------

    return {
        "agent": "route_agent",
        "status": "success",

        "zone_id": row["zone_id"],

        "location": {
            "latitude": safe_float(row["latitude"]),
            "longitude": safe_float(row["longitude"])
        },

        "route_metrics": {
            "distance_km": safe_float(row["distance_km"]),
            "travel_time_hr": safe_float(row["travel_time_hr"]),
            "fuel_l": safe_float(row["fuel_l_proxy"])
        },

        "route_components": {
            "efficiency_score": safe_float(
                row["efficiency_score_v2"]
            ),
            "depth_safety_score": safe_float(
                row["depth_safety_score"]
            ),
            "constraint_safety_score": safe_float(
                row["constraint_safety_score"]
            ),
            "current_safety_score": safe_float(
                row["current_safety_score"]
            ),
            "route_safety_score": safe_float(
                row["route_safety_score"]
            )
        },

        "route_score": safe_float(
            row["final_route_score"]
        ),

        "confidence": safe_float(
            row["final_route_confidence"],
            decimals=2
        ),

        "adjusted_route_score": safe_float(
            row["adjusted_final_route_score"]
        ),

        "data_quality": {
            "features_available": int(
                row["route_core_available"]
            ),
            "features_total": 5,
            "completeness": safe_float(
                row["route_data_completeness"],
                decimals=2
            )
        }
    }


# ------------------------------------------------------------
# TEST
# ------------------------------------------------------------

test_zone = route_features_df["zone_id"].iloc[0]

route_json = get_route_conditions(
    test_zone
)

print(
    json.dumps(
        route_json,
        indent=2
    )
)
```

```python
# ============================================================
# STEP 24: ROUTE AGENT QUERY INTERFACE
# ============================================================

def route_agent_query(
    query_type,
    zone_id=None,
    available_fuel_l=None,
    top_k=5
):
    """
    Common structured interface for the Route Agent.

    Supported query types:
    - zone
    - distance
    - time
    - fuel
    - shortest
    - closest
    - safest
    - lowest_fuel
    - fastest
    - fuel_feasibility
    - route
    """

    query_type = str(query_type).lower().strip()

    supported_queries = [
        "zone",
        "distance",
        "time",
        "fuel",
        "shortest",
        "closest",
        "safest",
        "lowest_fuel",
        "fastest",
        "fuel_feasibility",
        "route"
    ]

    # --------------------------------------------------------
    # Validate query type
    # --------------------------------------------------------

    if query_type not in supported_queries:
        return {
            "agent": "route_agent",
            "status": "error",
            "error": {
                "code": "unsupported_query_type",
                "message": (
                    "Supported query types: "
                    + ", ".join(supported_queries)
                )
            }
        }

    # --------------------------------------------------------
    # Single-zone query
    # --------------------------------------------------------

    if query_type in [
        "zone",
        "distance",
        "time",
        "fuel",
        "safest",
        "route"
    ]:

        if zone_id is None:
            return {
                "agent": "route_agent",
                "status": "error",
                "error": {
                    "code": "missing_zone_id",
                    "message": (
                        "zone_id is required "
                        "for this query."
                    )
                }
            }

        result = get_route_conditions(zone_id)

        if result.get("status") == "error":
            return result

        # Specific metric response
        if query_type == "distance":
            result = {
                "agent": "route_agent",
                "status": "success",
                "zone_id": result["zone_id"],
                "distance_km": (
                    result["route_metrics"]["distance_km"]
                )
            }

        elif query_type == "time":
            result = {
                "agent": "route_agent",
                "status": "success",
                "zone_id": result["zone_id"],
                "travel_time_hr": (
                    result["route_metrics"]["travel_time_hr"]
                )
            }

        elif query_type == "fuel":
            result = {
                "agent": "route_agent",
                "status": "success",
                "zone_id": result["zone_id"],
                "fuel_l": (
                    result["route_metrics"]["fuel_l"]
                )
            }

        return result

    # --------------------------------------------------------
    # Shortest / closest
    # --------------------------------------------------------

    if query_type in ["shortest", "closest"]:

        data = route_features_df.dropna(
            subset=["distance_km"]
        ).copy()

        row = data.loc[
            data["distance_km"].idxmin()
        ]

        return {
            "agent": "route_agent",
            "status": "success",
            "query": {
                "type": query_type
            },
            "result": {
                "zone_id": row["zone_id"],
                "distance_km": round(
                    float(row["distance_km"]),
                    4
                )
            }
        }

    # --------------------------------------------------------
    # Fastest
    # --------------------------------------------------------

    if query_type == "fastest":

        data = route_features_df.dropna(
            subset=["travel_time_hr"]
        ).copy()

        row = data.loc[
            data["travel_time_hr"].idxmin()
        ]

        return {
            "agent": "route_agent",
            "status": "success",
            "query": {
                "type": "fastest"
            },
            "result": {
                "zone_id": row["zone_id"],
                "travel_time_hr": round(
                    float(row["travel_time_hr"]),
                    4
                )
            }
        }

    # --------------------------------------------------------
    # Lowest fuel
    # --------------------------------------------------------

    if query_type == "lowest_fuel":

        data = route_features_df.dropna(
            subset=["fuel_l_proxy"]
        ).copy()

        row = data.loc[
            data["fuel_l_proxy"].idxmin()
        ]

        return {
            "agent": "route_agent",
            "status": "success",
            "query": {
                "type": "lowest_fuel"
            },
            "result": {
                "zone_id": row["zone_id"],
                "fuel_l": round(
                    float(row["fuel_l_proxy"]),
                    4
                )
            }
        }

    # --------------------------------------------------------
    # Fuel feasibility
    # --------------------------------------------------------

    if query_type == "fuel_feasibility":

        if zone_id is None:
            return {
                "agent": "route_agent",
                "status": "error",
                "error": {
                    "code": "missing_zone_id",
                    "message": (
                        "zone_id is required "
                        "for fuel feasibility."
                    )
                }
            }

        if available_fuel_l is None:
            return {
                "agent": "route_agent",
                "status": "error",
                "error": {
                    "code": "missing_fuel",
                    "message": (
                        "available_fuel_l is required."
                    )
                }
            }

        row = route_features_df[
            route_features_df["zone_id"] == zone_id
        ]

        if row.empty:
            return {
                "agent": "route_agent",
                "status": "error",
                "error": {
                    "code": "zone_not_found",
                    "message": (
                        f"Zone {zone_id} not found."
                    )
                }
            }

        row = row.iloc[0]

        required_fuel = row["fuel_l_proxy"]

        if pd.isna(required_fuel):
            return {
                "agent": "route_agent",
                "status": "error",
                "error": {
                    "code": "fuel_unavailable",
                    "message": (
                        "Fuel requirement unavailable."
                    )
                }
            }

        required_fuel = float(required_fuel)

        return {
            "agent": "route_agent",
            "status": "success",
            "query": {
                "type": "fuel_feasibility",
                "zone_id": zone_id,
                "available_fuel_l": float(
                    available_fuel_l
                )
            },
            "result": {
                "zone_id": zone_id,
                "required_fuel_l": round(
                    required_fuel,
                    4
                ),
                "available_fuel_l": round(
                    float(available_fuel_l),
                    4
                ),
                "reachable": (
                    float(available_fuel_l)
                    >= required_fuel
                )
            }
        }

    return {
        "agent": "route_agent",
        "status": "error",
        "error": {
            "code": "unhandled_query",
            "message": "Query could not be processed."
        }
    }


print("✅ Route Agent query interface created")
```

```python
# ============================================================
# STEP 25: ROUTE AGENT PARAMETER VALIDATION
# ============================================================

def validate_route_request(request):
    """
    Validate a structured Route Agent request.

    Returns:
        {
            "valid": True
        }
        or
        {
            "valid": False,
            "error": {...}
        }
    """

    # --------------------------------------------------------
    # 1. Request must be a dictionary
    # --------------------------------------------------------

    if not isinstance(request, dict):
        return {
            "valid": False,
            "error": {
                "code": "invalid_request",
                "message": "Request must be a JSON object."
            }
        }

    # --------------------------------------------------------
    # 2. query_type required
    # --------------------------------------------------------

    query_type = request.get("query_type")

    supported = [
        "zone",
        "distance",
        "time",
        "fuel",
        "shortest",
        "closest",
        "safest",
        "lowest_fuel",
        "fastest",
        "fuel_feasibility",
        "route"
    ]

    if query_type is None:
        return {
            "valid": False,
            "error": {
                "code": "missing_query_type",
                "message": "query_type is required."
            },
            "supported_query_types": supported
        }

    query_type = str(query_type).lower().strip()

    # --------------------------------------------------------
    # 3. Supported query type
    # --------------------------------------------------------

    if query_type not in supported:
        return {
            "valid": False,
            "error": {
                "code": "unsupported_query_type",
                "message": (
                    f"Unsupported query type: {query_type}"
                )
            },
            "supported_query_types": supported
        }

    # --------------------------------------------------------
    # 4. Zone-dependent queries
    # --------------------------------------------------------

    zone_queries = [
        "zone",
        "distance",
        "time",
        "fuel",
        "safest",
        "route",
        "fuel_feasibility"
    ]

    if query_type in zone_queries:

        if not request.get("zone_id"):
            return {
                "valid": False,
                "error": {
                    "code": "missing_zone_id",
                    "message": (
                        "zone_id is required for "
                        f"{query_type} query."
                    )
                }
            }

    # --------------------------------------------------------
    # 5. Fuel feasibility validation
    # --------------------------------------------------------

    if query_type == "fuel_feasibility":

        fuel = request.get("available_fuel_l")

        if fuel is None:
            return {
                "valid": False,
                "error": {
                    "code": "missing_available_fuel",
                    "message": (
                        "available_fuel_l is required "
                        "for fuel feasibility query."
                    )
                }
            }

        if isinstance(fuel, bool):
            return {
                "valid": False,
                "error": {
                    "code": "invalid_available_fuel",
                    "message": (
                        "available_fuel_l must be numeric."
                    )
                }
            }

        try:
            fuel = float(fuel)
        except (TypeError, ValueError):
            return {
                "valid": False,
                "error": {
                    "code": "invalid_available_fuel",
                    "message": (
                        "available_fuel_l must be numeric."
                    )
                }
            }

        if fuel < 0:
            return {
                "valid": False,
                "error": {
                    "code": "invalid_available_fuel",
                    "message": (
                        "available_fuel_l cannot be negative."
                    )
                }
            }

    # --------------------------------------------------------
    # 6. Valid request
    # --------------------------------------------------------

    return {
        "valid": True
    }


print("✅ Route Agent parameter validation created")
```

```python
# ============================================================
# STEP 26: FINAL VALIDATED ROUTE AGENT WRAPPER
# ============================================================

def run_route_agent(request):
    """
    Final structured Route Agent interface.

    Request:
        JSON-compatible dictionary

    Response:
        JSON-compatible Route Agent response
    """

    # --------------------------------------------------------
    # 1. Validate request
    # --------------------------------------------------------

    validation = validate_route_request(request)

    if not validation["valid"]:
        return {
            "agent": "route_agent",
            "status": "error",
            "error": validation["error"],
            **(
                {
                    "supported_query_types":
                    validation["supported_query_types"]
                }
                if "supported_query_types" in validation
                else {}
            )
        }

    # --------------------------------------------------------
    # 2. Normalize query type
    # --------------------------------------------------------

    query_type = (
        str(request["query_type"])
        .lower()
        .strip()
    )

    # --------------------------------------------------------
    # 3. Execute Route Agent query
    # --------------------------------------------------------

    response = route_agent_query(
        query_type=query_type,
        zone_id=request.get("zone_id"),
        available_fuel_l=request.get(
            "available_fuel_l"
        ),
        top_k=request.get("top_k", 5)
    )

    # --------------------------------------------------------
    # 4. Return structured response
    # --------------------------------------------------------

    return response


print("✅ Validated Route Agent wrapper created")
```

```python
# ============================================================
# STEP 27: NATURAL-LANGUAGE ROUTE QUERY PARSER
# ============================================================

import re


def parse_route_query(query):
    """
    Convert a natural-language routing query into
    a structured Route Agent request.
    """

    # --------------------------------------------------------
    # 1. Validate query
    # --------------------------------------------------------

    if not isinstance(query, str) or not query.strip():
        return {
            "status": "error",
            "error": "invalid_query",
            "message": "Query must be a non-empty string."
        }

    q = query.lower().strip()

    # --------------------------------------------------------
    # 2. Detect zone ID
    # --------------------------------------------------------

    zone_match = re.search(
        r"\b(?:pfz|grid)\d+\b",
        q,
        re.IGNORECASE
    )

    detected_zone = (
        zone_match.group(0).upper()
        if zone_match
        else None
    )

    # --------------------------------------------------------
    # 3. Detect fuel value
    # --------------------------------------------------------

    fuel_match = re.search(
        r"(\d+(?:\.\d+)?)\s*(?:litres?|liters?|l)\b",
        q,
        re.IGNORECASE
    )

    detected_fuel = (
        float(fuel_match.group(1))
        if fuel_match
        else None
    )

    # --------------------------------------------------------
    # 4. Determine query type
    # --------------------------------------------------------

    # Fuel feasibility
    if (
        ("reach" in q or "reaching" in q)
        and ("fuel" in q)
    ):
        query_type = "fuel_feasibility"

    # Shortest / closest
    elif (
        "shortest route" in q
        or "shortest" in q
    ):
        query_type = "shortest"

    elif (
        "closest" in q
        or "nearest" in q
    ):
        query_type = "closest"

    # Lowest fuel
    elif (
        "least fuel" in q
        or "lowest fuel" in q
        or "minimum fuel" in q
        or "min fuel" in q
    ):
        query_type = "lowest_fuel"

    # Travel time
    elif (
        "travel time" in q
        or "how long" in q
        or "time" in q
    ):
        query_type = "time"

    # Distance
    elif (
        "how far" in q
        or "distance" in q
        or "far" in q
    ):
        query_type = "distance"

    # Safest route
    elif (
        "safest route" in q
        or "safest" in q
        or "safe route" in q
    ):
        query_type = "safest"

    # Fastest route
    elif (
        "fastest route" in q
        or "lowest travel time" in q
        or "quickest route" in q
    ):
        query_type = "fastest"

    # Fuel requirement
    elif (
        "fuel" in q
        and (
            "need" in q
            or "require" in q
            or "required" in q
        )
    ):
        query_type = "fuel"

    # General route request
    elif "route" in q:
        query_type = "route"

    else:
        return {
            "status": "error",
            "error": "unsupported_query",
            "message": (
                "Could not determine the routing query type."
            )
        }

    # --------------------------------------------------------
    # 5. Build structured request
    # --------------------------------------------------------

    request = {
        "query_type": query_type
    }

    if detected_zone is not None:
        request["zone_id"] = detected_zone

    if detected_fuel is not None:
        request["available_fuel_l"] = detected_fuel

    return {
        "status": "success",
        **request
    }


print("✅ Natural-language Route Query Parser created")
```

```python
# ============================================================
# STEP 28: NATURAL LANGUAGE ROUTE AGENT
# ============================================================

def ask_route_agent(query):
    """
    Final natural-language Route Agent interface.

    Flow:
        Natural language
            ↓
        Query parser
            ↓
        Request validation
            ↓
        Route Agent
            ↓
        Structured JSON
    """

    # --------------------------------------------------------
    # 1. Parse natural-language query
    # --------------------------------------------------------

    parsed_request = parse_route_query(query)

    if parsed_request.get("status") != "success":
        error_response = {
            "agent": "route_agent",
            "status": "error",
            "error": {
                "code": parsed_request.get(
                    "error",
                    "query_parse_error"
                ),
                "message": parsed_request.get(
                    "message",
                    "Could not understand the routing query."
                )
            }
        }

        return {
            "answer": error_response["error"]["message"],
            "data": error_response
        }

    # --------------------------------------------------------
    # 2. Remove parser-only status field
    # --------------------------------------------------------

    agent_request = {
        key: value
        for key, value in parsed_request.items()
        if key != "status"
    }

    # --------------------------------------------------------
    # 3. Run validated Route Agent
    # --------------------------------------------------------

    response = run_route_agent(agent_request)

    # --------------------------------------------------------
    # 4. Generate simple human-readable answer
    # --------------------------------------------------------

    if response.get("status") == "success":

        query_type = agent_request.get(
            "query_type"
        )

        if query_type == "distance":
            answer = (
                f"Distance to {response['zone_id']} is "
                f"{response['distance_km']} km."
            )

        elif query_type == "time":
            answer = (
                f"Travel time to {response['zone_id']} is "
                f"{response['travel_time_hr']} hours."
            )

        elif query_type == "fuel":
            answer = (
                f"Estimated fuel requirement for "
                f"{response['zone_id']} is "
                f"{response['fuel_l']} litres."
            )

        elif query_type == "fuel_feasibility":
            result = response.get("result", {})

            answer = (
                f"{result.get('zone_id')} is "
                f"{'reachable' if result.get('reachable') else 'not reachable'} "
                f"with {result.get('available_fuel_l')} litres of fuel."
            )

        elif query_type in [
            "shortest",
            "closest",
            "fastest",
            "lowest_fuel"
        ]:
            result = response.get("result", {})

            answer = (
                f"Best matching zone is "
                f"{result.get('zone_id')}."
            )

        elif query_type == "safest":
            result = response.get("result", {})

            answer = (
                f"The highest-scoring route is "
                f"{result.get('zone_id')} with a "
                f"route score of {result.get('route_score')}."
            )

        else:
            answer = (
                f"Route information generated successfully "
                f"for {response.get('zone_id')}."
            )

    else:
        answer = response.get(
            "error",
            {}).get(
                "message",
                "Route Agent could not process the query."
            )

    # --------------------------------------------------------
    # 5. Return answer + machine-readable data
    # --------------------------------------------------------

    return {
        "answer": answer,
        "data": response
    }


print("✅ Final natural-language Route Agent created")
```

```python
# ============================================================
# STEP 35: CREATE FINAL M3 predict.py
# ============================================================

predict_code = r'''
import os
import joblib
import pandas as pd
import numpy as np

MODEL_PATH = os.path.join(
    os.path.dirname(__file__),
    "route_model.pkl"
)

model_artifact = joblib.load(MODEL_PATH)


def safe_float(value, decimals=4):
    if pd.isna(value):
        return None
    return round(float(value), decimals)


def predict_route(row):
    """
    Generate the final structured Route Agent output.

    The supplied dataset route_score is NOT used
    for prediction.
    """

    efficiency = row.get("efficiency_score_v2")
    route_safety = row.get("route_safety_score")

    components = []

    if pd.notna(efficiency):
        components.append(
            (
                float(efficiency),
                model_artifact["route_score_weights"]["efficiency"]
            )
        )

    if pd.notna(route_safety):
        components.append(
            (
                float(route_safety),
                model_artifact["route_score_weights"]["safety"]
            )
        )

    if not components:
        return {
            "agent": "route_agent",
            "status": "error",
            "error": {
                "code": "insufficient_route_data",
                "message": "Insufficient routing information."
            }
        }

    total_weight = sum(
        weight for score, weight in components
    )

    route_score = sum(
        score * weight
        for score, weight in components
    ) / total_weight

    route_score = float(
        np.clip(route_score, 0, 1)
    )

    available = sum(
        pd.notna(row.get(feature))
        for feature in [
            "distance_km",
            "travel_time_hr",
            "fuel_l_proxy",
            "water_depth_m",
            "current_speed_ms"
        ]
    )

    confidence = available / 5.0

    return {
        "agent": "route_agent",
        "status": "success",

        "zone_id": str(row["zone_id"]),

        "location": {
            "latitude": safe_float(row.get("latitude")),
            "longitude": safe_float(row.get("longitude"))
        },

        "route_metrics": {
            "distance_km": safe_float(
                row.get("distance_km")
            ),
            "travel_time_hr": safe_float(
                row.get("travel_time_hr")
            ),
            "fuel_l": safe_float(
                row.get("fuel_l_proxy")
            )
        },

        "route_score": round(
            route_score,
            4
        ),

        "confidence": round(
            confidence,
            2
        ),

        "data_quality": {
            "features_available": int(available),
            "features_total": 5,
            "completeness": round(
                confidence,
                2
            )
        }
    }
'''

with open("predict.py", "w") as f:
    f.write(predict_code)

print("✅ predict.py created")
```
